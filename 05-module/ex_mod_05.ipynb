```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df = pd.read_excel('../data/processed/relatorio_final.xlsx')
```

```python
df.head()
```

```python
df.info()
```

```python
plt.hist(df['Salario'], bins=20, color='orchid')
plt.title("Salary Distribution in the Company")
plt.xlabel("Salary in BRL")
plt.ylabel("Frequency")
plt.grid(True)
plt.show()
```

O histograma demonstra que a maior concentração de salários está entre aproximadamente R$ 4.000 e R$ 9.000, indicando que a maioria dos funcionários se encontra em faixas salariais intermediárias. Observa-se também menor frequência de salários muito elevados, acima de R$ 12.000, sugerindo que cargos com remuneração mais alta representam uma parcela reduzida do conjunto de funcionários. A distribuição apresenta dispersão moderada, permitindo identificar diferenças salariais dentro da empresa.

# =========================================
# Exercise 2 — Comparação Categórica
# =========================================

```python
department_counts = df['Departamento'].value_counts()
```

```python
print(department_counts)
```

```python
plt.bar(
    department_counts.index,
    department_counts.values,
    color='purple'
)
plt.title("Number of Employees by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees")

plt.show()
```

O gráfico de barras evidencia que o departamento de Vendas possui o maior número de funcionários, seguido pelo setor de TI. O departamento de RH apresenta a menor quantidade de colaboradores entre os três setores analisados. Essa distribuição indica uma maior concentração de profissionais em áreas diretamente relacionadas às operações e atividades estratégicas da empresa, enquanto setores administrativos possuem menor representatividade.

# =========================================
# Exercise 3 — Correlação   
# =========================================

```python
plt.scatter(
    df['Idade'],
    df['Salario'],
    color='green',
    alpha=0.6
)

plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")

plt.show()
```

```python
# Age of Clara using label
df['Idade'].max()
```

O gráfico de dispersão indica que não existe uma relação linear forte entre idade e salário, uma vez que os pontos estão distribuídos de forma relativamente espalhada ao longo do gráfico. Apesar disso, observa-se maior concentração de funcionários entre aproximadamente 25 e 45 anos, com salários variando em diferentes faixas. A visualização também evidencia a presença de alguns salários mais elevados distribuídos entre diferentes idades, sugerindo que fatores além da idade podem influenciar a remuneração.

# =========================================
# Exercise 4 — Mapeamento Rápido via Pandas
# =========================================

```python
average_salary_by_age = df.groupby('Idade')['Salario'].mean()

average_salary_by_age.plot(
    kind='line',
    marker='o'
)

plt.title("Average Salary by Age")
plt.xlabel("Age")
plt.ylabel("Average Salary")

plt.grid(True)

plt.show()
```

O gráfico de linha apresenta oscilações na média salarial entre as diferentes idades, sem demonstrar um crescimento contínuo ao longo do tempo. Em algumas faixas etárias observa-se aumento expressivo da média salarial, enquanto em outras há redução significativa, indicando variabilidade nos dados. Essas variações podem estar relacionadas à quantidade de funcionários em determinadas idades ou à presença de salários mais altos e mais baixos influenciando a média.

# =========================================
# Exercise 5 — Proporções Relativas
# =========================================

```python
department_counts = df['Departamento'].value_counts()

plt.pie(
    department_counts.values,
    labels=department_counts.index,
    autopct='%1.1f%%'
)

plt.title("Employee Division by Department")

plt.show()
```

O gráfico de pizza evidencia a participação proporcional de cada departamento no total de funcionários da empresa. O setor de Vendas representa aproximadamente metade do quadro funcional, com cerca de 49,6% dos colaboradores, seguido pelo departamento de TI com 34,0%. Já o setor de RH apresenta a menor participação, correspondendo a aproximadamente 16,4% do total. A visualização facilita a compreensão da distribuição relativa de funcionários entre os departamentos.
